```python
import numpy as np
import pandas as pd
```

```python
# import numpy as np
students = pd.read_csv('https://raw.githubusercontent.com/kiat/Elements-of-Data-Analytics/refs/heads/main/datasets/Librarie/Students.csv')
books = pd.read_csv('https://raw.githubusercontent.com/kiat/Elements-of-Data-Analytics/refs/heads/main/datasets/Librarie/Books.csv')
borrow = pd.read_csv('https://raw.githubusercontent.com/kiat/Elements-of-Data-Analytics/refs/heads/main/datasets/Librarie/Borrow.csv')
students.head()
```

# Query 1: List all students names and ids of Students of Year 2026.

```python
students[students['Year'] == 2026][['Name', 'SID']]
```

# Query 2: Find all books in the "Technology" category

```python
books[books['Category'] == "Technology"][["BID", "Title", "Author"]]
```

# Query 3: Find names of students from the “Computer Science” department

```python
students[students["Department"] == "Computer Science"]["Name"]
```

# Query 4: Find titles of all books borrowed by students

```python
pd.merge(borrow, books, on ='BID', how = 'left')["Title"].drop_duplicates().dropna()
```

# Query 5: Find names of students who have borrowed at least one book

```python
pd.merge(borrow, students, on = "SID", how = "left")["Name"].drop_duplicates().sort_values()
```

# Query 6: Find students who have borrowed "Technology" books

```python
student_ids = pd.merge(borrow, books[books["Category"] == "Technology"], on = "BID", how = "inner")["SID"].drop_duplicates()
pd.merge(student_ids, students, on = "SID", how = "left")["Name"]
```

# Query 7: Find students who have not returned their books

```python
pd.merge(borrow[borrow["ReturnDate"].isnull()], students, on = "SID", how = "left")["Name"].drop_duplicates()
```

# Query 10: Find top-10 students based on borrowing count



```python
pd.merge(borrow, students, on = "SID", how = "left")["Name"].value_counts().head(10)
```

# Query 11: Find departments sorted by total number of borrowed books

```python
pd.merge(borrow, students, on = "SID", how = "left")["Department"].value_counts()
```

# Query 12: Find number of borrowers per department

```python
pd.merge(borrow, students, on = "SID", how = "left").groupby("Department")["SID"].nunique()
```

# Query 13: Find departments sorted by total number of borrowed books

```python
pd.merge(borrow, students, on = "SID", how = "left")["Department"].value_counts()
```

# Query 14: Count how many times each book has been borrowed (including books never borrowed)

```python
pd.merge(books, borrow, on="BID", how="outer").groupby("BID")["SID"].count()
```

# Query 15: Count number of borrowed books per department (including departments with zero borrows)

```python
pd.merge(borrow, students, on = "SID", how = "outer").groupby("Department")["BID"].count().sort_values(ascending = False)
```

# Query 16: Count total number of students and total number of borrows per book category

```python
pd.merge(borrow, books, on = "BID", how = "outer").groupby("Category")["SID"].count()
```
